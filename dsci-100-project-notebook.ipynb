#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
stat_table_numeric_only <- wmm_data_tidy |> # select only numerical columns
    select(-tourney_finish)

stat_table_mean <- stat_table_numeric_only |>  # calculate mean
    map_dfr(mean, na.rm = TRUE)

stat_table_max <- stat_table_numeric_only |> # find max values
    map_dfr(max, na.rm = TRUE)

stat_table_min <- stat_table_numeric_only |> # find min values
    map_dfr(min, na.rm = TRUE)

stat_table <- bind_rows(.id = "statistic",   # combine mean, min and max into one table, and also add the statistic column
                        mean = stat_table_mean,
                        min  = stat_table_min,
                        max  = stat_table_max)
stat_table
```

It’s important to gauge the relationship between a numerical and a categorical variable before we build our model. I’ve decided to use a boxplot for this. Here we can visually assess the relationship between regional win percentage and what level was reached at a tournament. At a first glance, we can see a potential association between higher regional win percentage and higher tournament levels. However, the boxes overlap quite a bit, especially for the earlier finishes, so regional win percentage alone probably can’t tell every level apart.

```r
#Considering I have two numerical variables that should be used to predict a categorical variable, I think it makes sense to create 2 boxplots to study the distribtuion of each numerical variable in the context of a categorical variable.

finish_levels <- c("opening_round_loss", "first_round_loss", "second_round_loss", #logical order of tournament finish levels
                   "top_16_loss", "top_8_loss", "top_4_loss", "top_2_loss", "champ")

wmm_plot_data <- wmm_data_tidy |> #converting tourney_finish chr column to fct. This way the boxplot will be ordered from opening_round_loss to champ
    mutate(tourney_finish = factor(tourney_finish, levels = finish_levels))

reg_boxplot <- wmm_plot_data |>
    ggplot(aes(x = tourney_finish, y = reg_wins_pct)) +
    geom_boxplot() +
    labs(x = "Tournament Finish",
         y = "Regional Win Percentage",
         title = "Regional Win Percentage by Tournament Finish") +
    theme(text = element_text(size = 15),
          axis.text.x = element_text(angle = 45, hjust = 1))

reg_boxplot
```

Looking at this boxplot, we can visually assess the relationship between conference win percentage and tournament finish. It seems like higher conference win percentages are somewhat associated with deeper tournament finishes, though the trend is less consistent than the regional win percentage we plotted earlier. The boxes overlap a lot, so conference win percentage alone probably can’t separate every level, which is part of why we will be using two predictor variables. It is also worth noting that opening_round_loss contains only one team with a conference win percentage, so that category isn’t very informative.

```r
conf_box <- wmm_plot_data |>
    ggplot(aes(x = tourney_finish, y = conf_wins_pct)) +
    geom_boxplot() +
    labs(x = "Tournament Finish",
         y = "Conference Win Percentage",
         title = "Conference Win Percentage by Tournament Finish") +
    theme(text = element_text(size = 15),
          axis.text.x = element_text(angle = 45, hjust = 1))
conf_box
```
